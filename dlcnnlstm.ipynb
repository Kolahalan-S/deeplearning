```python
import numpy as np
import random
import matplotlib.pyplot as plt
import tensorflow as tf

from tensorflow.keras.datasets import fashion_mnist
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Model
from tensorflow.keras.layers import (
    Input, Conv2D, MaxPooling2D, GlobalAveragePooling2D,
    Dense, Embedding, LSTM, Add, Dropout
)


# 1. Load and preprocess images
(x_train, y_train), (x_test, y_test) = fashion_mnist.load_data()

x_train = x_train.astype("float32")[..., np.newaxis] / 255.0
x_test = x_test.astype("float32")[..., np.newaxis] / 255.0

class_names = [
    "t-shirt", "trouser", "pullover", "dress", "coat",
    "sandal", "shirt", "sneaker", "bag", "ankle boot"
]

# Use a smaller dataset to train faster
np.random.seed(42)
random.seed(42)

train_idx = np.random.choice(len(x_train), 4000, replace=False)
test_idx = np.random.choice(len(x_test), 500, replace=False)

train_images = x_train[train_idx]
test_images = x_test[test_idx]

# 2. Generate captions from image labels
templates = [
    "a photo of a {}",
    "this picture shows a {}",
    "an image of a {}"
]

def make_caption(label):
    name = class_names[label]
    article = "an" if name[0] in "aeiou" else "a"
    caption = random.choice(templates).format(name)
    caption = caption.replace("a " + name, article + " " + name)
    return "startseq " + caption + " endseq"

train_captions = [make_caption(y_train[i]) for i in train_idx]
test_captions = [make_caption(y_test[i]) for i in test_idx]

# 3. Convert words into numbers
tokenizer = Tokenizer()
tokenizer.fit_on_texts(train_captions)

vocab_size = len(tokenizer.word_index) + 1
max_len = max(len(c.split()) for c in train_captions)

sequences = tokenizer.texts_to_sequences(train_captions)

# Create (image, partial caption) -> next word training examples
X_img, X_seq, Y = [], [], []

for img, seq in zip(train_images, sequences):
    for i in range(1, len(seq)):
        X_img.append(img)
        X_seq.append(pad_sequences([seq[:i]], maxlen=max_len)[0])
        Y.append(seq[i])

X_img = np.array(X_img)
X_seq = np.array(X_seq)
Y = np.array(Y)

# 4. Build CNN encoder
image_input = Input(shape=(28, 28, 1))

cnn = Conv2D(32, 3, activation="relu", padding="same")(image_input)
cnn = MaxPooling2D()(cnn)
cnn = Conv2D(64, 3, activation="relu", padding="same")(cnn)
cnn = MaxPooling2D()(cnn)
cnn = GlobalAveragePooling2D()(cnn)
image_features = Dense(128, activation="relu")(cnn)

# 5. Build LSTM decoder
caption_input = Input(shape=(max_len,))
text = Embedding(vocab_size, 128, mask_zero=True)(caption_input)
text = LSTM(128)(text)

# Combine image features and caption features
combined = Add()([image_features, text])
combined = Dense(128, activation="relu")(combined)
output = Dense(vocab_size, activation="softmax")(combined)

model = Model([image_input, caption_input], output)
model.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

model.summary()

# 6. Train the model
history = model.fit(
    [X_img, X_seq], Y,
    epochs=10,
    batch_size=64,
    validation_split=0.1
)

# Plot training loss
plt.plot(history.history["loss"], label="Training Loss")
plt.plot(history.history["val_loss"], label="Validation Loss")
plt.title("CNN-LSTM Caption Generation")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.legend()
plt.savefig("caption_training_loss.png")
plt.show()

# 7. Generate captions word by word
index_to_word = {
    index: word for word, index in tokenizer.word_index.items()
}

def generate_caption(image):
    caption = "startseq"

    for _ in range(max_len):
        seq = tokenizer.texts_to_sequences([caption])
        seq = pad_sequences(seq, maxlen=max_len)

        prediction = model.predict(
            [image[np.newaxis, ...], seq], verbose=0
        )

        next_word_id = int(np.argmax(prediction))
        next_word = index_to_word.get(next_word_id, "")

        if next_word in ("", "endseq"):
            break

        caption += " " + next_word

    return caption.replace("startseq", "").strip()

# 8. Display predictions on test images
plt.figure(figsize=(12, 8))

for i in range(6):
    image = test_images[i]
    predicted = generate_caption(image)
    actual = class_names[y_test[test_idx[i]]]

    plt.subplot(2, 3, i + 1)
    plt.imshow(image.squeeze(), cmap="gray")
    plt.title(f"Actual: {actual}\nCaption: {predicted}")
    plt.axis("off")

plt.tight_layout()
plt.savefig("caption_predictions.png")
plt.show()
```
